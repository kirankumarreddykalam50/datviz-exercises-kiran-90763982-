```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Set default Plotly theme
import plotly.io as pio
pio.templates.default = "plotly_white"

# 1. Load Dataset
df = pd.read_csv(r"C:\Users\kiran\OneDrive\Desktop\dataviz_exercises\dataviz-exercises-kirankumar-reddy\final_project\data\Video_Games.csv")

# 2. Data Preprocessing & Cleaning
df_clean = df.copy()

# Remove index column if present
if 'index' in df_clean.columns:
    df_clean = df_clean.drop(columns=['index'])

# Convert User_Score to numeric (coercing 'tbd' and string NaNs to float)
df_clean['User_Score'] = pd.to_numeric(df_clean['User_Score'], errors='coerce')

# Convert Year_of_Release to Int64 (nullable integer)
df_clean['Year_of_Release'] = pd.to_numeric(df_clean['Year_of_Release'], errors='coerce')

# Scale Critic_Score to 0-10 scale for direct comparison with User_Score
df_clean['Critic_Score_Scaled'] = df_clean['Critic_Score'] / 10.0

# Calculate Score Gap (User Score - Critic Score Scaled)
df_clean['Score_Gap'] = df_clean['User_Score'] - df_clean['Critic_Score_Scaled']

print(f"Dataset successfully cleaned: {len(df_clean)} records remaining.")
```

```python
# Display core summary statistics
print("--- Sales & Metrics Overview ---")
print(df_clean[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales', 'Critic_Score', 'User_Score']].describe())

# Check missing values
missing = df_clean.isnull().sum()
missing_pct = (missing / len(df_clean)) * 100
print("\n--- Missing Value Breakdown (%) ---")
print(pd.DataFrame({'Missing Count': missing, 'Percentage (%)': missing_pct.round(2)}))
```

```python
# Aggregate regional sales by genre
genre_sales = df_clean.groupby('Genre')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum().reset_index()

# Melt for Plotly
genre_sales_melted = genre_sales.melt(id_vars=['Genre'], var_name='Region', value_name='Sales')
genre_sales_melted['Region'] = genre_sales_melted['Region'].str.replace('_Sales', '')

fig1 = px.bar(
    genre_sales_melted,
    x='Genre',
    y='Sales',
    color='Region',
    title='<b>Regional Sales Breakdown across Gaming Genres</b><br><sup>Japan dominates Role-Playing market share while North America leads Action and Sports</sup>',
    labels={'Sales': 'Sales (Millions of Units)', 'Genre': 'Game Genre'},
    color_discrete_sequence=px.colors.qualitative.Safe,
    barmode='stack'
)
fig1.update_layout(xaxis_tickangle=-45, legend_title_text='Region')
fig1.show()
```

```python
# Map platforms into major ecosystems
def categorize_platform(plat):
    if plat in ['PS', 'PS2', 'PS3', 'PS4', 'PSP', 'PSV']:
        return 'PlayStation'
    elif plat in ['X360', 'XB', 'XOne']:
        return 'Xbox'
    elif plat in ['N64', 'GC', 'Wii', 'WiiU', 'GB', 'GBA', 'DS', '3DS', 'NES', 'SNES']:
        return 'Nintendo'
    elif plat == 'PC':
        return 'PC'
    else:
        return 'Others'

df_clean['Ecosystem'] = df_clean['Platform'].apply(categorize_platform)

# Group sales by year and ecosystem
eco_year = df_clean[df_clean['Year_of_Release'].between(1990, 2016)].groupby(['Year_of_Release', 'Ecosystem'])['Global_Sales'].sum().reset_index()

fig2 = px.line(
    eco_year,
    x='Year_of_Release',
    y='Global_Sales',
    color='Ecosystem',
    title='<b>Console Ecosystem Market Shifts Over Time (1990–2016)</b><br><sup>Nintendo peaked in mid-2000s with Wii/DS while PlayStation maintained steady multi-generational market share</sup>',
    labels={'Global_Sales': 'Global Sales (Millions)', 'Year_of_Release': 'Release Year'},
    color_discrete_sequence=px.colors.qualitative.Bold
)
fig2.show()
```

```python
# Filter rows with complete score and rating data
ratings_df = df_clean.dropna(subset=['Critic_Score_Scaled', 'User_Score', 'Rating', 'Global_Sales'])
ratings_df = ratings_df[ratings_df['Rating'].isin(['E', 'E10+', 'T', 'M'])]

fig3 = px.scatter(
    ratings_df,
    x='Critic_Score_Scaled',
    y='User_Score',
    color='Rating',
    size='Global_Sales',
    hover_name='Name',
    title='<b>Critic vs. User Score Consensus by ESRB Content Rating</b><br><sup>Bubble size indicates Global Sales (in Millions)</sup>',
    labels={'Critic_Score_Scaled': 'Critic Score (0–10 Scale)', 'User_Score': 'User Score (0–10 Scale)'},
    color_discrete_sequence=px.colors.qualitative.Dark24,
    opacity=0.6
)
# Add identity diagonal reference line
fig3.add_trace(go.Scatter(x=[0, 10], y=[0, 10], mode='lines', line=dict(dash='dash', color='gray'), name='Ideal Consensus (1:1)'))
fig3.show()
```

```python
# Top 8 publishers by total sales
top_pubs = df_clean.groupby('Publisher')['Global_Sales'].sum().nlargest(8).index
pub_df = df_clean[df_clean['Publisher'].isin(top_pubs)]

pub_genre = pub_df.groupby(['Publisher', 'Genre'])['Global_Sales'].sum().reset_index()

fig4 = px.bar(
    pub_genre,
    x='Publisher',
    y='Global_Sales',
    color='Genre',
    title='<b>Top 8 Publisher Revenue Breakdown by Genre</b><br><sup>EA relies heavily on Sports, while Nintendo exhibits portfolio diversification across Platformers and RPGs</sup>',
    labels={'Global_Sales': 'Global Sales (Millions)', 'Publisher': 'Publisher'},
    color_discrete_sequence=px.colors.qualitative.Prism
)
fig4.update_layout(xaxis_tickangle=-30)
fig4.show()
```

```python
rev_df = df_clean.dropna(subset=['Critic_Count', 'User_Count', 'Score_Gap', 'Global_Sales'])
rev_df = rev_df[(rev_df['User_Count'] > 50) & (rev_df['Critic_Count'] > 10)]

fig5 = px.scatter(
    rev_df,
    x='Critic_Count',
    y='User_Count',
    color='Score_Gap',
    size='Global_Sales',
    hover_name='Name',
    color_continuous_scale='Viridis',
    title='<b>Review Volume Density vs. Score Divergence</b><br><sup>Higher review volumes coincide with blockbuster titles and narrowed rating disparities</sup>',
    labels={'Critic_Count': 'Number of Critic Reviews', 'User_Count': 'Number of User Reviews', 'Score_Gap': 'Score Gap (User - Critic)'}
)
fig5.show()
```

```python
annual_metrics = df_clean[df_clean['Year_of_Release'].between(1990, 2016)].groupby('Year_of_Release').agg(
    Total_Releases=('Name', 'count'),
    Avg_Sales_Per_Game=('Global_Sales', 'mean')
).reset_index()

fig6 = make_subplots(specs=[[{"secondary_y": True}]])

fig6.add_trace(
    go.Bar(x=annual_metrics['Year_of_Release'], y=annual_metrics['Total_Releases'], name="Total Titles Released", marker_color='#2b5c8f'),
    secondary_y=False
)

fig6.add_trace(
    go.Scatter(x=annual_metrics['Year_of_Release'], y=annual_metrics['Avg_Sales_Per_Game'], name="Avg Sales / Game (M)", line=dict(color='#d95f02', width=3)),
    secondary_y=True
)

fig6.update_layout(
    title_text="<b>Annual Release Volume vs. Average Revenue per Title (1990–2016)</b><br><sup>As release volume surged in 2008–2011, average sales per individual title significantly diluted</sup>"
)
fig6.update_xaxes(title_text="Release Year")
fig6.update_yaxes(title_text="<b>Total Titles Released</b>", secondary_y=False)
fig6.update_yaxes(title_text="<b>Average Sales per Title ($M)</b>", secondary_y=True)
fig6.show()
```

```python
jp_west = df_clean[df_clean['Year_of_Release'].between(1990, 2016)].groupby('Year_of_Release').agg(
    JP_Total=('JP_Sales', 'sum'),
    West_Total=('NA_Sales', lambda x: x.sum()) # NA + EU
).reset_index()

# Add EU to West
eu_temp = df_clean[df_clean['Year_of_Release'].between(1990, 2016)].groupby('Year_of_Release')['EU_Sales'].sum().values
jp_west['West_Total'] = jp_west['West_Total'] + eu_temp
jp_west['JP_Share_%'] = (jp_west['JP_Total'] / (jp_west['JP_Total'] + jp_west['West_Total'])) * 100

fig7 = px.area(
    jp_west,
    x='Year_of_Release',
    y='JP_Share_%',
    title='<b>Japan Regional Market Share Concentration Trend (%)</b><br><sup>Japan held over 30% global market share in the 1990s before shifting toward Western dominance in the HD console era</sup>',
    labels={'JP_Share_%': 'Japan Share of Combined NA/EU/JP Sales (%)', 'Year_of_Release': 'Release Year'},
    color_discrete_sequence=['#7570b3']
)
fig7.show()
```

```python
esrb_era = df_clean.dropna(subset=['Rating', 'Year_of_Release'])
esrb_era = esrb_era[esrb_era['Rating'].isin(['E', 'E10+', 'T', 'M']) & esrb_era['Year_of_Release'].between(1995, 2016)]

esrb_era['Era'] = pd.cut(
    esrb_era['Year_of_Release'],
    bins=[1994, 2001, 2008, 2017],
    labels=['Early Era (1995-2001)', 'Mid Era (2002-2008)', 'Modern Era (2009-2016)']
)

fig8 = px.histogram(
    esrb_era,
    x='Rating',
    color='Era',
    barmode='group',
    title='<b>ESRB Rating Class Distribution Across Industry Generations</b><br><sup>Mature (M) rated titles grew steadily during the modern HD hardware era</sup>',
    labels={'count': 'Number of Titles', 'Rating': 'ESRB Content Rating'},
    color_discrete_sequence=px.colors.qualitative.Dark2
)
fig8.show()
```

```python
dev_stats = df_clean.groupby('Developer').filter(lambda x: len(x) >= 15)
dev_summary = dev_stats.groupby('Developer').agg(
    Avg_Critic_Score=('Critic_Score', 'mean'),
    Total_Blockbusters=('Global_Sales', lambda x: (x >= 2.0).sum()),
    Total_Games=('Global_Sales', 'count')
).reset_index().dropna()

fig9 = px.scatter(
    dev_summary,
    x='Avg_Critic_Score',
    y='Total_Blockbusters',
    size='Total_Games',
    hover_name='Developer',
    title='<b>Developer Hit Rate (Titles > 2M Sales) vs. Average Critic Rating</b><br><sup>Top developers with high critic averages consistently produce high-selling blockbuster titles</sup>',
    labels={'Avg_Critic_Score': 'Average Critic Score (0–100)', 'Total_Blockbusters': 'Blockbusters (> 2M Sales)'},
    color_discrete_sequence=['#e7298a']
)
fig9.show()
```

```python
ps_df = df_clean[df_clean['Platform'].isin(['PS2', 'PS3', 'PS4'])]
ps_genre = ps_df.groupby(['Platform', 'Genre'])['Global_Sales'].sum().reset_index()

fig10 = px.bar(
    ps_genre,
    x='Platform',
    y='Global_Sales',
    color='Genre',
    title='<b>PlayStation Hardware Generation Evolution: Sales Distribution by Genre</b><br><sup>Action titles gained substantial market share from PS2 to PS4 while Sports remained constant</sup>',
    labels={'Global_Sales': 'Global Sales (Millions)', 'Platform': 'PlayStation Generation'},
    color_discrete_sequence=px.colors.qualitative.Alphabet
)
fig10.show()
```
